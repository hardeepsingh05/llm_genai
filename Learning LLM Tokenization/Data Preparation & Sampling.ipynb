## Reading in a short story as text sample into Python

## Step 1: Creating Tokens

```python
# Taking input of the 'Verdict.txt' file
with open('the-verdict.txt','r',encoding = 'utf-8') as f:
    raw_text = f.read()

print("Total number of character:",len(raw_text))
print(raw_text[:99])
```

```python
# Now let's process the 'raw_text' and preform tokenization
import re
preprocessed_txt = re.split(r'([,.:;?_!"()\']|--|\s)',raw_text)
preprocessed_txt = [item.strip() for item in preprocessed_txt if item.strip()]
print(preprocessed_txt[0:30])
```

```python
print(len(preprocessed_txt))
```

## Step 2: Creating Token IDs

```python
# Getting a unique token count
all_words =  sorted(set(preprocessed_txt))
vocab_size = len(all_words)

print(vocab_size)
```

```python
# Making a dictionary for storing the tokens as keys and token number as value of key
vocab = {token:i for i,token in enumerate(all_words)}
```

```python
# Printing the first 50 items of this vocabulary 
for i,item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break
```

```python
# let's define a class called "Simple Tokenizer" in which we will define the encoder and decoder methods

class SimpleTokenizerV1:
    # Defining constructor with variable
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self,text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)',text)

        preprocessed = [item.strip() for item in preprocessed if item.strip()]

        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self,ids):
        text = " ".join([self.int_to_str[id] for id in ids])

        # Replace the spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])',r'\1',text)
        return text
        
```

```python
# Now let's use our tokenizer class to check some outputs
tokenizer = SimpleTokenizerV1(vocab)

text  ="""It's the last he painted, you know,"
           Mrs. Gisburn said with pardonable pride."""

ids = tokenizer.encode(text)
print(ids)
```

```python
# let's use our decoder
tokenizer.decode(ids)
```

```python
# Second Version of our vocabulary in order to update it for 'UNKNOWN' tokens and "ENDOFTEXT" token in order to mark the end of one text and beginning of the another.
class SimpleTokenizerV2:
    # Defining constructor with variable
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self,text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)',text)

        preprocessed = [item.strip() for item in preprocessed if item.strip()]

        ########  Additional part ###########
        preprocessed = [item if item in self.str_to_int else "<|unk|>" for item in preprocessed]
        ########  Additional part ###########
        
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self,ids):
        text = " ".join([self.int_to_str[id] for id in ids])

        # Replace the spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])',r'\1',text)
        return text
         
```

```python
# Try to use the updated tokenizer "V2"
tokenizer = SimpleTokenizerV2(vocab)
t1 = "Hello, do you like tea?"
t2 = "In the sunlit terraces of the palace."

text = " <|endoftext|> ".join((t1,t2))

token_ids = tokenizer.encode(text)
tokenizer.decode(token_ids)
```

```python
# Importing & Installing the tiktoken library for having encoder & decoder
import importlib
import tiktoken
```

```python
# Storing tokenizer to variable
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
# Encoding the text
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
    "of someunknownPlace."
)
ids = tokenizer.encode(text, allowed_special = {"<|endoftext|>"})
print(ids)
```

```python
# Decoding the text
strings = tokenizer.decode(ids)
print(strings)
```
